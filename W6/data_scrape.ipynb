```python
from psaw import PushshiftAPI
import pandas as pd

api = PushshiftAPI()

# Fetch 500 recent submissions from r/worldnews
gen = api.search_submissions(
    subreddit='worldnews',
    filter=['title', 'url', 'score', 'created_utc'],
    limit=500,
    sort='desc',
    sort_type='created_utc'
)

posts = []
for post in gen:
    posts.append({
        'title': post.title,
        'url': post.url,
        'score': post.score
    })

df = pd.DataFrame(posts)
print(df.head())
```

```python
import requests

API_KEY = "abfefee8a01f45cf85e24448b3c8c448"  # Get from https://newsapi.org/
url = f"https://newsapi.org/v2/everything"
params = {
    'q': 'worldnews',                     # all topics
    'sortBy': 'publishedAt',
    'pageSize': 100,             # max allowed
    'page': 1,
    'apiKey': API_KEY
}

all_articles = []

for page in range(1, 6):  # pages 1 to 5 → max 500 articles
    params['page'] = page
    response = requests.get(url, params=params)
    
    if response.status_code != 200:
        print(f"Error on page {page}: {response.json()}")
        break
        
    data = response.json()
    articles = data.get('articles', [])
    if not articles:
        break
    all_articles.extend(articles)
    print(f"Fetched {len(articles)} articles from page {page}")

print(f"Total articles collected: {len(all_articles)}")
```

```python
import requests
import time

API_KEY = "abfefee8a01f45cf85e24448b3c8c448"  # Replace with your actual API key
all_articles = []

# We'll fetch 10 pages x 100 articles = 1000 articles
total_pages = 10
page_size = 100

for page in range(1, total_pages + 1):
    print(f"Fetching page {page}...")
    url = (
        "https://newsapi.org/v2/everything"
        "?q=worldnews"
        "&sortBy=publishedAt"
        f"&pageSize={page_size}"
        f"&page={page}"
        f"&apiKey={API_KEY}"
    )
    
    response = requests.get(url)
    if response.status_code != 200:
        print(f"Error: {response.status_code} - {response.text}")
        break
        
    data = response.json()
    if data['status'] != 'ok':
        print(f"API Error: {data.get('message', 'Unknown error')}")
        break
        
    articles = data.get('articles', [])
    if not articles:
        print("No more articles. Stopping.")
        break
        
    all_articles.extend(articles)
    print(f"Fetched {len(articles)} articles (Total: {len(all_articles)})")
    
    # Be respectful: 1-second delay between requests
    time.sleep(1)

# Save or process the 1000 articles
print(f"\nTotal articles collected: {len(all_articles)}")
for i, article in enumerate(all_articles[:5], 1):  # Show first 5
    print(f"{i}. {article['title']}\n   {article['url']}\n")
```
