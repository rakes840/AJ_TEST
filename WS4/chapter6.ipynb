```python

# ALL-IN-ONE CELL: Build pipeline, cross-validate, fit, save to pickle, download link, load & predict.

import numpy as np
import pickle
import nltk
from sklearn.base import TransformerMixin, BaseEstimator
from sklearn.feature_extraction import DictVectorizer
from sklearn.naive_bayes import MultinomialNB  # You can switch to BernoulliNB if using binary features
from sklearn.pipeline import Pipeline
from sklearn.model_selection import cross_val_score, StratifiedKFold
from sklearn.metrics import f1_score
from pathlib import Path

# Optional (Jupyter environment only): provide a clickable download link after saving the pickle
try:
    from IPython.display import FileLink, display
    _HAS_IPYTHON = True
except Exception:
    _HAS_IPYTHON = False


# --------- SAFETY CHECKS ----------
if 'new_tweets' not in globals() or 'y_true' not in globals():
    raise ValueError("Please define 'new_tweets' (list of texts) and 'y_true' (labels) above before running this cell.")
if not isinstance(new_tweets, (list, tuple)) or not isinstance(y_true, (list, tuple, np.ndarray)):
    raise TypeError("'new_tweets' must be a list/tuple of strings and 'y_true' must be a list/tuple/array of labels.")
if len(new_tweets) != len(y_true):
    raise ValueError(f"Length mismatch: len(new_tweets)={len(new_tweets)} vs len(y_true)={len(y_true)}.")
if len(set(y_true)) < 2:
    raise ValueError("Need at least two distinct classes in y_true for classification.")
# if 'X_test' not in globals():
#     raise ValueError("Please define 'X_test' (list of texts for prediction) above before running this cell.")

# --------- NLTK RESOURCES ----------
# Ensure 'punkt' tokenizer is available
try:
    nltk.data.find('tokenizers/punkt')
except LookupError:
    nltk.download('punkt')

class NLTKBOW(BaseEstimator, TransformerMixin):
    """
    Transformer that converts raw text into feature dicts using NLTK tokenization.
    - Produces unigram counts and optional bigram counts.
    - Set lowercase=True to normalize case.
    - Set use_bigrams=True to include bigrams.
    """
    def __init__(self, lowercase=True, use_bigrams=True):
        self.lowercase = lowercase
        self.use_bigrams = use_bigrams

    def fit(self, X, y=None):
        # No fitting needed for BoW creation
        return self

    def transform(self, X):
        feature_dicts = []
        for text in X:
            if not isinstance(text, str):
                # Convert non-string entries (e.g., None, numbers) to strings safely
                text = "" if text is None else str(text)

            t = text.lower() if self.lowercase else text
            tokens = nltk.word_tokenize(t)

            feats = {}
            # Unigram counts
            for tok in tokens:
                key = f"UNI_{tok}"
                feats[key] = feats.get(key, 0) + 1

            # Optional bigrams
            if self.use_bigrams and len(tokens) > 1:
                for i in range(len(tokens) - 1):
                    bg = f"{tokens[i]}_{tokens[i+1]}"
                    key = f"BIG_{bg}"
                    feats[key] = feats.get(key, 0) + 1

            feature_dicts.append(feats)
        return feature_dicts

# --------- PIPELINE DEFINITION ----------
# MultinomialNB works well with count-based features
pipeline = Pipeline([
    ('nltk_bow', NLTKBOW(lowercase=True, use_bigrams=True)),
    ('vectorizer', DictVectorizer(sparse=True)),
    ('clf', MultinomialNB())
])

# --------- CROSS-VALIDATION ----------
# Use macro F1 for multi-class or imbalanced datasets
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
scores = cross_val_score(pipeline, new_tweets, y_true, cv=cv, scoring='f1_macro')
print(f"Cross-val F1-macro (5-fold): mean={np.mean(scores):.3f}, std={np.std(scores):.3f}, folds={np.round(scores, 3)}")

# --------- FIT THE MODEL ----------
pipeline.fit(new_tweets, y_true)

# --------- SAVE TO PICKLE ----------
pickle_path = Path("tweet_nb_pipeline.pkl")
with open(pickle_path, "wb") as f:
    pickle.dump(pipeline, f)
print(f"✅ Saved fitted pipeline to: {pickle_path.resolve()}")

# Optional: show a download link if running in Jupyter
if _HAS_IPYTHON:
    display(FileLink(pickle_path.name))
```

```python
# --------- LOAD FROM PICKLE ----------
with open(pickle_path, "rb") as f:
    loaded_pipeline = pickle.load(f)

X_test = ["I love python coding","Java is complicated language compare to other languages",
          "The data types inside the python is List and tuples"]

# --------- PREDICT ON NEW DATA ----------
preds = loaded_pipeline.predict(X_test)
# Optional: predicted probabilities (works for NB classifiers)
try:
    proba = loaded_pipeline.predict_proba(X_test)
except Exception:
    proba = None

print("\nPredictions on X_test:")
for i, text in enumerate(X_test):
    if proba is not None:
        # Show top class and its probability
        top_class = np.argmax(proba[i])
        top_prob = proba[i][top_class]
        print(f"- {repr(text)} -> pred={preds[i]} (top_prob={top_prob:.3f})")
    else:
        print(f"- {repr(text)} -> pred={preds[i]}")
```
